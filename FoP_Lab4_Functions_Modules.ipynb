## <font color="#C00000">3.3 Tuples: packing and unpacking</font>

<font color="#C00000">`return total, average` packs two values into a tuple; `t, a = f()` unpacks it. Swapping: `a, b = b, a`.</font>

````python
def calculate_stats(numbers):
    total = sum(numbers)
    average = total / len(numbers)

    # PACKING: Python packs these two values into a single tuple: (total, average)
    return total, average

scores = [85, 90, 95]

# First, see what the raw return value looks like (a packed tuple)
raw_result = calculate_stats(scores)
print(f"Packed tuple: {raw_result} (Type: {type(raw_result)})")

# UNPACKING: Assign the returned tuple's elements directly to two variables
t, a = calculate_stats(scores)
print(f"Unpacked -> Total: {t}, Average: {a}")
````

````python
a = "First"
b = "Second"

# Right side packs (b, a) into a temporary tuple ("Second", "First")
# Left side immediately unpacks that tuple into variables a and b
a, b = b, a

print(f"a: {a}")
print(f"b: {b}")
````

## <font color="#C00000">3.4 Scope</font>

<font color="#C00000">A variable created inside a function is **local** – it exists only while the function runs. Variables created at the top level of the script are **global**; a function can read them, but to *assign* to a global it must declare `global name`. Prefer passing values as parameters and returning results; use globals only for constants (e.g. `PASSING_GRADE = 60`) or simple counters.</font>

````python
# Global variable (accessible anywhere in the file)
PASSING_GRADE = 60

def check_result(score):
    # 'status' is a local variable; it exists only during this call
    if score >= PASSING_GRADE:
        status = "Pass"
    else:
        status = "Fail"
    return status

print(check_result(75))  # Output: Pass

# Trying to access a local variable outside the function raises an error:
# print(status)  # NameError: name 'status' is not defined
````

````python
counter = 0  # Global counter

def increment_bad():
    # Without 'global', Python treats 'counter' as local, causing an UnboundLocalError
    # if you try: counter += 1
    counter = 10  # This only creates a local variable named 'counter'
    print(f"Inside increment_bad: {counter}")

def increment_good():
    global counter  # Tells Python to target the top-level 'counter'
    counter += 1
    print(f"Inside increment_good: {counter}")

increment_bad()
print(f"Global counter after bad: {counter}")   # Output: 0 (unchanged)

increment_good()
print(f"Global counter after good: {counter}")  # Output: 1 (modified)
````

````python
# --- AVOID: Hidden side effects using 'global' ---
current_balance = 100

def withdraw_global(amount):
    global current_balance
    current_balance -= amount

withdraw_global(30)


# --- PREFERRED: Pure function taking inputs and returning results ---
def withdraw(balance, amount):
    """Takes balance, returns updated balance without touching global state."""
    return balance - amount

initial_balance = 100
new_balance = withdraw(initial_balance, 30)
print(f"Remaining balance: ${new_balance}")  # Output: Remaining balance: $70
````

## <font color="#C00000">3.6 Methods</font>

<font color="#C00000">A **method** is a function that belongs to an object and is called with a dot: `"hcmiu".upper()`, `"Lab".center(20, "*")`, `prices.append(9.5)`, `prices.sort()`.</font>

````python
# .upper() converts all characters to uppercase
university = "hcmiu"
loud_name = university.upper()

print(f"Original: {university}")   # hcmiu (unchanged)
print(f"Upper:    {loud_name}")    # HCMIU

# .center(width, fillchar) centers the text within padding
header = "Lab"
banner = header.center(20, "*")

print(banner)  # ********Lab*********
````

````python
prices = [12.0, 5.5, 8.0]

# .append(value) adds an item to the end of the list
prices.append(9.5)
print(f"After append: {prices}")   # [12.0, 5.5, 8.0, 9.5]

# .sort() orders elements in ascending order in-place
prices.sort()
print(f"After sort:   {prices}")   # [5.5, 8.0, 9.5, 12.0]
````

````python
# [ADDED] worked example – run this cell
calls = 0                               # global counter

def add(a, b):
    global calls
    calls += 1
    result = a + b                      # local variable
    return result

add(2, 3); add(4, 5)
print("calls =", calls)                 # 2
# print(result)  -> NameError: result only exists inside add()
````

### <font color="#0070C0">🔵 [REVISED] Task 4 · Level 2 ⭐⭐ (Basic) — Analysis → Algorithm</font>

- <font color="#0070C0">**① Understand the problem** — Input: number of flips (default 100). Output: (heads, tails) and their percentages.</font>

<pre style="color:#0070C0;background:#EAF3FB;padding:8px 12px;border-left:4px solid #0070C0">FUNCTION toss(flips=100):

````python
# ✍️ Task 8 – your code
# Write your code below, then run the cell (Shift+Enter).
import random

total_games_played = 0

def roll(n, sides=6):
    global total_games_played
    total_games_played += 1

    rolls = []
    for _ in range(n):
        rolls.append(random.randint(1, sides))

    print(" dice roll results ".upper().center(25, "*"))
    print(f"Rolls: {rolls}")
    return min(rolls), max(rolls), sum(rolls) / n

print("Roll 3 dice with 6 sides")
minimum, maximum, average = roll(3)
print(f"Min: {minimum} | Max: {maximum} | Avg: {average:.2f}")
print(f"Games played: {total_games_played}")

print()

print("Roll 5 dice with 20 sides")
minimum, maximum, average = roll(5, sides=20)
print(f"Min: {minimum} | Max: {maximum} | Avg: {average:.2f}")
print(f"Games played: {total_games_played}")
````

````python
# Extended
import random
import math

RISK_FREE_RATE = 0.03

def add_investment(name, value, volatility, *tags):
    """Return an investment as a tuple."""
    return name, value, volatility, tags

def simulate_yearly_return(initial, volatility):
    """Return the average, best and worst values from 100 simulations."""
    results = []

    for _ in range(100):
        r = random.gauss(0.07, volatility)
        final_value = initial * (1 + r)
        results.append(final_value)

    average = math.fsum(results) / len(results)
    best = max(results)
    worst = min(results)

    return average, best, worst

def calculate_risk_reward(value, volatility):
    """Return the risk level and Sharpe ratio for a return rate."""
    if volatility < 0.1:
        risk = "LOW"
    elif volatility < 0.3:
        risk = "MEDIUM"
    elif volatility < 0.6:
        risk = "HIGH"
    else:
        risk = "EXTREME"

    if volatility == 0:
        sharpe = math.nan
    else:
        sharpe = (value - RISK_FREE_RATE) / volatility

    return risk, sharpe

def generate_report(portfolio, risk_tolerance=0.15):
    """Return the portfolio performance and risk report."""
    total_initial = math.fsum(item[1] for item in portfolio)

    if risk_tolerance <= 0.1:
        tolerance_label = "LOW"
    elif risk_tolerance < 0.3:
        tolerance_label = "MODERATE"
    else:
        tolerance_label = "HIGH"

    lines = [
        "financial portfolio report".upper(),
        "─" * 50,
        f"Client Risk Tolerance: {tolerance_label} ({risk_tolerance:.1%})",
        f"Total Initial Value  : ${total_initial:,.2f}",
        "Simulated 1-Year Growth (random market):"
    ]

    final_values = []
    weighted_risks = []

    for index, investment in enumerate(portfolio, start=1):
        name, initial, volatility, tags = investment

        final_value, best, worst = simulate_yearly_return(initial, volatility)        r = (final_value - initial) / initial
        risk, sharpe = calculate_risk_reward(r, volatility)

        final_values.append(final_value)
        weighted_risks.append(initial * volatility)

        warning = ""
        if volatility > risk_tolerance:
            warning = " [WARNING]"

        sharpe_text = "N/A" if math.isnan(sharpe) else f"{sharpe:.2f}"

        lines.append(
            f"{index}. {name.upper()} | "
            f"${initial:,.0f} → ${final_value:,.0f} | "
            f"Risk: {risk:<7} | Sharpe: {sharpe_text}{warning}"
        )

    total_final = math.fsum(final_values)
    gain = total_final - total_initial
    gain_percent = gain / total_initial * 100

    overall_risk = math.fsum(weighted_risks) / total_initial
    overall_label, _ = calculate_risk_reward(0, overall_risk)

    lines.append(f"Final Portfolio Value: ${total_final:,.0f}")
    lines.append(
        f"Projected Gain/Loss  : "
        f"{'+' if gain >= 0 else '-'}${abs(gain):,.0f} "
        f"({gain_percent:+.2f}%)"
    )
    lines.append(
        f"Overall Risk Score   : {overall_risk:.2f} ({overall_label})"
    )

    if overall_risk > risk_tolerance:
        highest_risk = portfolio[0]

        for investment in portfolio:
            if investment[2] > highest_risk[2]:
                highest_risk = investment

        lines.append(
            f"Recommendation: Reduce {highest_risk[0]} exposure. "
            "Consider rebalancing."
        )
    else:
        lines.append(
            "Recommendation: Maintain the current portfolio."
        )

    return "\n".join(lines)

def main():
    """Read investments and display the portfolio report."""
    portfolio = []

    print("=== Financial Portfolio Analyzer ===")

    while True:
        name = input(
            f"Enter investment {len(portfolio) + 1} name: "
        ).strip()

        if name == "":
            break

        value = float(input("Initial value: "))
        volatility = float(input("Volatility (e.g. 0.25): "))
        tags_text = input("Tags (comma-separated): ")

        tags = []
        for tag in tags_text.split(","):
            tag = tag.strip()
            if tag:
                tags.append(tag)

        investment = add_investment(name, value, volatility, *tags)
        portfolio.append(investment)

    if not portfolio:
        return

    tolerance = input(
        "Your risk tolerance (0.1 = low, 0.3 = high) "
        "[default 0.15]: "
    ).strip()

    if tolerance == "":
        report = generate_report(portfolio)
    else:
        report = generate_report(
            portfolio,
            risk_tolerance=float(tolerance)
        )

    print(report)


main()
````

````python
# ✍️ Task 10 – your code
# Write your code below, then run the cell (Shift+Enter).
import random
import string

def make_password(length=12, use_digits=True, use_symbols=True):
    """Return a password containing every selected character group."""
    groups = [string.ascii_lowercase, string.ascii_uppercase]

    if use_digits:
        groups.append(string.digits)
    if use_symbols:
        groups.append(string.punctuation)

    chars = [random.choice(group) for group in groups]
    all_chars = "".join(groups)

    while len(chars) < length:
        chars.append(random.choice(all_chars))

    random.shuffle(chars)
    return "".join(chars)


def strength(pw):
    """Return the password strength score and label."""
    has_lower = False
    has_upper = False
    has_digit = False
    has_symbol = False

    for char in pw:
        if char.islower():
            has_lower = True
        elif char.isupper():
            has_upper = True
        elif char.isdigit():
            has_digit = True
        elif char in string.punctuation:
            has_symbol = True

    score = int(has_lower) + int(has_upper) + int(has_digit) + int(has_symbol)

    if score <= 1:
        label = "weak"
    elif score <= 3:
        label = "medium"
    else:
        label = "strong"
    return score, label

pw = make_password()
print(pw, strength(pw))

pw = make_password(8, True, False)
print(pw, strength(pw))

pw = make_password(length=10, use_digits=False, use_symbols=True)
print(pw, strength(pw=pw))
````
